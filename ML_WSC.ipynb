```python
#Regularisation
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
dataset=pd.read_csv("L1_L2_Regularization_Dataset.csv")
dataset
```

```python
#To visualize corellation we use heatmap
sns.heatmap(dataset.corr(),annot=True)
plt.show()
```

```python
x=dataset.iloc[:,:-1]
y=dataset["Target"]
```

```python
sc=StandardScaler()
sc.fit(x)
x=pd.DataFrame(sc.transform(x),columns=x.columns)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression , Lasso , Ridge
lr=LinearRegression()
la=Lasso(alpha=1)    #Alpha----> penalty
ri=Ridge()
```

```python
lr.fit(x_train,y_train)
lr.score(x_test,y_test)
```

```python
la.fit(x_train,y_train)
la.score(x_test,y_test)
```

```python
ri.fit(x_train,y_train)
ri.score(x_test,y_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
import numpy as np 
```

```python
mean_squared_error(y_test,lr.predict(x_test))
```

```python
mean_absolute_error(y_test,lr.predict(x_test))
```

```python
mean_squared_error(y_test,la.predict(x_test))
```

```python
mean_absolute_error(y_test,la.predict(x_test))
```

```python
mean_squared_error(y_test,ri.predict(x_test))
```

```python
mean_absolute_error(y_test,ri.predict(x_test))
```

```python
#Comparing coeeficients 
```
