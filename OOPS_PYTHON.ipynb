```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def avg_marks(self):
        sum=0
        for i in self.marks:
            sum=sum+i
        avg=sum/len(self.marks)
        return avg
s1=Student('Nikhil',[98,99,97])
print(s1.avg_marks())
```

```python
#Static method 
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    
    @staticmethod
    def hello_():
        print('hello')
s1=Student('Nikhil',100)
s1.hello_()
```

```python
class Account:
    def __init__(self,balance,acc_no):
        self.balance=balance
        self.acc_no=acc_no
    def debit_cash(self,debit_amt):
        print('Amount debited')
        self.balance=(self.balance - (debit_amt))
        print('Remaining amount : ',self.balance)
    def credit_cash(self,credit_amt):
        print('Amount Created')
        self.balance=(self.balance + (credit_amt))
        print('Remaining amount : ',self.balance)
U1=Account(1000000,4567687)
U1.debit_cash(100000)
U1.debit_cash(100000)
U1.credit_cash(200000)
```

```python
#Inheritance ----->One one class derives the properties and methods of another class
class Car:
    colour='black'
    @staticmethod
    def start():
        print('Car Started')
    def stop():
        print('Car Stopped')
class Toyota_car(Car):
    def __init__(self,name):
        self.name=name
c1=Toyota_car('Bleno')
print(c1.start())
```

```python
print(c1.colour)
```

```python
class Car:
    @staticmethod
    def start():
        print('Car has started')
    @staticmethod
    def stop():
        print('Car has stopped')
class ToyotaCar(Car):
    def __init__(self,brand):
        self.brand=brand
class fortuner(ToyotaCar):
    def __init__(self,petrol):
        self.petrol=petrol
c1=fortuner(1000)
c1.start()
```

```python
#Super method 
class Car:
    def __init__(self,type):
        self.type=type
    @staticmethod
    def start():
        print("Car Started")
    @staticmethod
    def stop():
        print("Car Stopped")
class ToyotaCar(Car):
    def __init__(self,name,type):
        self.name=name
        super().__init__(type)
        super().start()
c1=ToyotaCar('Mecedis','Electic')
c1.type
```
