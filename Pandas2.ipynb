#Merge

```python
import pandas as pd 
```

```python
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4,5,6],"c":[15,16,17,18,19,20]})
pd.merge(var_1,var_2,on="a")
```

```python
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4,5,6],"c":[15,16,17,18,19,20]})
pd.merge(var_2,var_1,on="a")
```

```python
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4,5,9],"c":[15,16,17,18,19,20]})
pd.merge(var_2,var_1,on="a")
```

```python
#how parameter ------>Default is inner
pd.merge(var_1,var_2,how="inner")
```

```python
pd.merge(var_1,var_2,how="left")
```

```python
pd.merge(var_1,var_2,how="right")
```

```python
pd.merge(var_1,var_2,how="outer")
```

```python
#Indicator ------->To know what data are present absent 
pd.merge(var_1,var_2,how="outer",indicator=True)
```

```python
pd.merge(var_1,var_2,left_index=True,right_index=True)
```

```python
#To change name we use suffixes parameter
pd.merge(var_1,var_2,left_index=True,right_index=True,suffixes=["name","id"])
```

```python
#Concat 
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4,5,9],"c":[15,16,17,18,19,2]})
```

```python
import pandas as pd
s=pd.Series([1,2,3,4,5])
r=pd.Series([6,7,8,9,10])
pd.concat([s,r])
```

```python
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4,5,9],"b":[15,16,17,18,19,20]})
pd.concat([var_1,var_2])
```

```python
pd.concat([var_1,var_2],axis=1)
```

```python
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4],"b":[15,16,17,18]})
pd.concat([var_1,var_2],axis=1)
```

```python
#Join parameter ------inner or outer 
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4],"b":[15,16,17,18]})
pd.concat([var_1,var_2],axis=1,join="inner")
```

```python
#key parameter 
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4],"b":[15,16,17,18]})
pd.concat([var_1,var_2],axis=1,keys=["d1","d2"])
```

```python
var_1=pd.DataFrame({"a":[1,2,3,4,5,6],"b":[12,13,14,15,16,17]})
var_2=pd.DataFrame({"a":[1,2,3,4],"b":[15,16,17,18]})
pd.concat([var_1,var_2],axis=0,keys=["d1","d2"])
```

```python
#Merging 1d and 2d dataframes
var_1=pd.DataFrame({"a":[12,23,34,45,56]})
var_2=pd.DataFrame({"a":[12,23,34,45,56,78],"b":[21,42,32,65,76,89]})
pd.concat([var_1,var_2],axis=0)
```

```python
import pandas as pd
var_1=pd.DataFrame({"a":[1,2,3,4,5],"b":[6,7,8,9,10]})
var_2=pd.DataFrame({"c":[11,12,13,14,15],"d":[16,17,18,19,20]})
var_1.join(var_2)
```

```python
var_1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
var_2=pd.DataFrame({"C":[11,12,13],"D":[16,17,18]},["a","b","c"])
var_2.join(var_1)
```

```python
#how parameter
var_1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
var_2=pd.DataFrame({"C":[11,12,13],"D":[16,17,18]},["a","b","c"])
var_2.join(var_1,how="left")
```

```python
var_1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
var_2=pd.DataFrame({"C":[11,12,13],"D":[16,17,18]},["a","b","c"])
var_2.join(var_1,how="right")
```

```python
#lsuffix---->when 2 column names are same 
var_1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
var_2=pd.DataFrame({"C":[11,12,13],"B":[16,17,18]},["a","b","c"])
var_1.join(var_2,lsuffix="_12")
```

```python
var_1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
var_2=pd.DataFrame({"C":[11,12,13],"B":[16,17,18]},["a","b","c"])
var_1.join(var_2,rsuffix="_12")
```

```python
var_1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]},index=["a","b","c","d","e"])
var_2=pd.DataFrame({"C":[11,12,13],"B":[16,17,18]},["a","b","c"])
var_1.join(var_2,lsuffix="_123",rsuffix="_12")
```
