```python
from gensim.models import Word2Vec
import nltk
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
corpus = "The cat sat on the mat. The dog sat on the rug Cats and dogs are great pets. The quick brown fox jumps over the lazy dog. Foxes and dogs are wild animals sometimes."
```

```python
sentences=nltk.sent_tokenize(corpus)
```

```python
sentences
```

```python
lemmitizer=WordNetLemmatizer()
```

```python
import re 
clean_corpus=[]
for i in range(len(sentences)):
    review=re.sub('[^a-zA-Z]',' ',sentences[i])
    review=review.lower()
    review=review.split()
    review=[lemmitizer.lemmatize(word) for word in review if word not in set(stopwords.words('english'))]
    review=' '.join(review)
    clean_corpus.append(review)
```

```python
clean_corpus
```

```python
clean_corpus = [sentence.split() for sentence in clean_corpus]
```

```python
model=Word2Vec(
    sentences=clean_corpus,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    epochs=50
)
```

```python
vec=model.wv['dog']
```

```python
vec.shape
```

```python
print(model.wv.index_to_key)
```

```python
print(model.wv.most_similar("dog", topn=3))
```

```python
print(model.wv.similarity("cat", "dog"))
```

```python
print(model.wv.doesnt_match(["cat", "dog", "fox", "rug"]))
```
