```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 120

data = {
    "Student_ID": range(1, n+1),

    "Experience_Level": np.random.choice(
        ["Beginner", "Intermediate", "Advanced"], n
    ),

    "Education_Level": np.random.choice(
        ["School", "UG", "PG", "PhD"], n
    ),

    "Performance": np.random.choice(
        ["Low", "Medium", "High"], n
    ),

    "Satisfaction": np.random.choice(
        ["Very Low", "Low", "Medium", "High", "Very High"], n
    ),

    "Salary_Range": np.random.choice(
        ["0-3L", "3-6L", "6-10L", "10L+"], n
    )
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
b=["Experience_Level","Education_Level","Performance","Satisfaction","Salary_Range"]
ord_map = {
    "Experience_Level": {"Beginner": 0, "Intermediate": 1, "Advanced": 2},
    "Education_Level": {"School": 4, "UG": 5, "PG": 6, "PhD": 7},
    "Performance": {"Low": 8, "Medium": 9, "High": 10},
    "Satisfaction": {"Very Low": 0, "Low": 1, "Medium": 2, "High": 3, "Very High": 4},
    "Salary_Range": {"0-3L": 0, "3-6L": 1, "6-10L": 2, "10L+": 3}
}
for i in b:
    df[i+"_en"]=df[i].map(ord_map[i]).fillna(-1)
```
