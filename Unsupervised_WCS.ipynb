```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
dataset=pd.read_csv('iris_dataset.csv')
```

```python
dataset.head(3)
```

```python
dataset.drop(columns=['target','species'],inplace=True)
```

```python
dataset.head(3)
```

```python
sns.pairplot(data=dataset)
plt.plot()
```

```python
from sklearn.cluster import KMeans
```

```python
from sklearn.metrics import silhouette_score
```

```python
ss=[]
for i in range(2,21):
    kmc=KMeans(n_clusters=i)
    kmc.fit(dataset)
    ss.append(silhouette_score(dataset,kmc.labels_))
```

```python
sns.lineplot(x=[i for i in range(2,21)],y=ss)
plt.plot()
plt.xticks([i for i in range(2,21)])
plt.grid()
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
from sklearn.datasets import make_moons
```

```python
x,y=make_moons(n_samples=1000,noise=0.2)
```

```python
dict={"x1":x[:,0],"x2":x[:,1],"y":y}
```

```python
dataset=pd.DataFrame(dict)
```

```python
dataset.head(3)
```

```python
sns.scatterplot(x='x1',y='x2',data=dataset,hue='y')
plt.show()
```

```python
x_a=dataset.iloc[:,:-1]
```

```python
y_a=dataset['y']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_a,y_a,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.svm import SVC
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)
```

```python
dt.score(x_train,y_train)
```

```python
sv=SVC()
```

```python
sv.fit(x_train,y_train)
```

```python
sv.score(x_train,y_train),sv.score(x_test,y_test)
```

```python
gnb=GaussianNB()
gnb.fit(x_train,y_train)
```

```python
gnb.score(x_train,y_train),gnb.score(x_test,y_test)
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
vc=VotingClassifier([('dt1',DecisionTreeClassifier()),('SV1',SVC()),('GNB1',GaussianNB())],weights=[5,5,1])
```

```python
vc.fit(x_train,y_train)
```

```python
vc.score(x_train,y_train)*100,vc.score(x_test,y_test)*100
```

```python
#To check its working we can see like :
```

```python
dic={'dt':dt.predict(x_train),'sv':sv.predict(x_train),'gnb':gnb.predict(x_train),'y':vc.predict(x_train)}
```

```python
df=pd.DataFrame(dic)
```
