```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Step 1: Load dataset
df = pd.read_csv("smartwatch_health_dataset.csv")

# Step 2: Select features
X = df[["Heart_Rate", "Steps", "Sleep_Hours", "Calories_Burned", "Activity_Level"]]

# Step 3: Normalize data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Apply Isolation Forest
model = IsolationForest(contamination=0.07)  # ~7% anomalies
df["Anomaly"] = model.fit_predict(X_scaled)

# Convert: -1 → anomaly, 1 → normal
df["Anomaly"] = df["Anomaly"].map({1: 0, -1: 1})

# -------------------------------
# 📊 Graph 1: Heart Rate vs Steps
# -------------------------------
plt.figure()
plt.scatter(df["Steps"], df["Heart_Rate"], c=df["Anomaly"])
plt.title("Anomaly Detection (Heart Rate vs Steps)")
plt.xlabel("Steps")
plt.ylabel("Heart Rate")
plt.show()

# -------------------------------
# 📊 Graph 2: Sleep vs Calories
# -------------------------------
plt.figure()
plt.scatter(df["Sleep_Hours"], df["Calories_Burned"], c=df["Anomaly"])
plt.title("Anomaly Detection (Sleep vs Calories)")
plt.xlabel("Sleep Hours")
plt.ylabel("Calories Burned")
plt.show()

# -------------------------------
# 📊 Graph 3: Count of anomalies
# -------------------------------
plt.figure()
df["Anomaly"].value_counts().plot(kind="bar")
plt.title("Normal vs Anomaly Count")
plt.xlabel("0 = Normal, 1 = Anomaly")
plt.ylabel("Count")
plt.show()

# Print few anomaly rows
print("Detected Anomalies:")
print(df[df["Anomaly"] == 1].head())
```
