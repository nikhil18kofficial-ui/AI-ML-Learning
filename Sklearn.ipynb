```python
import pandas as pd 
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(data=iris.data,columns=iris.feature_names)
```

```python
#EDA
import matplotlib.pyplot as plt
df.head()
df.shape
```

```python
df.plot.scatter(x='petal length (cm)',y='sepal width (cm)')
```

```python
#Gives correlation matric
print(df.corr())
```

```python
print(df.describe())
```

```python
#Data Preprocessing 
y=df['petal length (cm)'].values.reshape(-1,1)    #Converts into 2d column array 
x=df['sepal width (cm)'].values.reshape(-1,1)
```

```python
#Train test split 
from sklearn.model_selection import train_test_split 
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2)
print(X_train)
```

```python
print(X_test)
```

```python
print(Y_train)
```

```python
print(Y_test)
```

```python
#Train the model (y=wx+b)
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(X_train,Y_train)
print(regressor.intercept_)     #Calculate the value of b
print(regressor.coef_)          #Calculate the value of w
```

```python
#Prediction 
score=regressor.predict([[4]])
print(score)
```

```python
y_pred=regressor.predict(X_test)
df_pred=pd.DataFrame({"Actual":Y_test.squeeze(),"Predicted":y_pred.squeeze()})
print(df_pred)
```

EVALUATE THE MODEL

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
import numpy as np 
mae=mean_absolute_error(Y_test,y_pred)
mse=mean_squared_error(Y_test,y_pred)
rmse=np.sqrt(mse)
print(mae)
print(mse)
print(rmse)
```
