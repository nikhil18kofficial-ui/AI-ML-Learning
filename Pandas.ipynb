```python
#merge two data frames - To merge two df then we need to keep one common column
import pandas as pd
var1=pd.DataFrame({"A":[1,2,3,4,5,6],"B":[7,8,9,10,11,12]})
var2=pd.DataFrame({"A":[1,2,3,4,5,6],"C":[13,14,15,16,17,18]})
pd.merge(var1,var2,on="A")
```

```python
#IF ONE DATA IS MISSED ONLY COMMON ELEMENTS WITLL GET PRIONTED
import pandas as pd
var1=pd.DataFrame({"A":[1,2,3,4,5,6],"B":[7,8,9,10,11,12]})
var2=pd.DataFrame({"A":[1,2,3,9,5,6],"C":[13,14,15,16,17,18]})
pd.merge(var1,var2,on="A")
```

```python
#How parameter-inner(default),left,right,outer
import pandas as pd
var1=pd.DataFrame({"A":[1,2,3,4,5,6],"B":[7,8,9,10,11,12]})
var2=pd.DataFrame({"A":[1,2,3,9,5,6],"C":[13,14,15,16,17,18]})
pd.merge(var1,var2,how="inner")
```

```python
import pandas as pd
var1=pd.DataFrame({"A":[1,2,3,4,5,6],"B":[7,8,9,10,11,12]})
var2=pd.DataFrame({"A":[1,2,3,9,5,6],"C":[13,14,15,16,17,18]})
pd.merge(var1,var2,how="left")
```
