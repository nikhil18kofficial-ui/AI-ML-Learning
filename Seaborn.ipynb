```python
import matplotlib.pyplot as plt 
import seaborn as sns 
import pandas as pd 
```

```python
var=[1,2,3,4,5,6]
var_1=[2,3,4,1,5,6]
x=pd.DataFrame({"var":var,"var_1":var_1})
sns.lineplot(x="var",y="var_1",data=x)
plt.show()
```

```python
x=pd.read_csv("attention.csv")
x
```

```python
sns.lineplot(x="Unnamed: 0",y="score",data=x)
plt.show()
```

```python
#hue parameter
sns.lineplot(x="Unnamed: 0",y="score",data=x,hue="attention")
```

```python
#Size parameter
sns.lineplot(x="Unnamed: 0",y="score",data=x,hue="attention",size=10,style="attention",palette="Accent")
plt.show()
```

```python
#Marker Parameter
sns.lineplot(x="Unnamed: 0",y="score",data=x,hue="attention",size=10,style="attention",palette="Accent",markers=["o",">"])
plt.show()
```

```python
#Dashes
sns.lineplot(x="Unnamed: 0",y="score",data=x,hue="attention",size=10,style="attention",palette="Accent",markers=["o",">"],dashes=False)
plt.show()
```
