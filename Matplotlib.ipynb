```python
#to change the properties of Text on pie chat we use textprops parameter in which dictiory should be passed
import matplotlib.pyplot as plt
x=[10,20,30,40]
y=["Python","Java","C","C++"]
c=["y","r","m","b"]
ex=[0.0,0.4,0.0,0.0]
plt.pie(x,labels=y,colors=c,autopct="%0.1f%%",explode=ex,startangle=45,shadow=True,textprops={"fontsize":15})
plt.show()
```

```python
#can also change the direction(Anti clock wise to clockwise counterclock parameter
import matplotlib.pyplot as plt
x=[10,20,30,40]
y=["Python","Java","C","C++"]
c=["y","r","m","b"]
ex=[0.0,0.4,0.0,0.0]
plt.pie(x,labels=y,colors=c,autopct="%0.1f%%",explode=ex,startangle=45,shadow=True,textprops={"fontsize":12},counterclock=False)
plt.show()
```

```python
#wedgeprops
import matplotlib.pyplot as plt
x=[10,20,30,40]
y=["Python","Java","C","C++"]
c=["y","r","m","b"]
ex=[0.0,0.4,0.0,0.0]
plt.pie(x,labels=y,colors=c,autopct="%0.1f%%",explode=ex,startangle=90,shadow=True,textprops={"fontsize":12},wedgeprops={"linewidth":10,"edgecolor":"m"})
plt.show()
```

```python
#Stem plot
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y)
plt.show()
```

```python
#linefmt parameter --- To change the solid line into dotted line
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y,linefmt=":")
plt.show()
```

```python
#markerfmt parameter---To change the marker and colorr of it
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y,linefmt=":",markerfmt="r*")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y,linefmt=":",markerfmt="ro")
plt.show()
```

```python
#To change the bottom line we use bottom parameter
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y,linefmt=":",markerfmt="ro",bottom=5)
plt.show()
```

```python
#To change the color of base we use basefmt
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y,linefmt=":",markerfmt="ro",bottom=5,basefmt="g")
plt.show()
```

```python
#Orientation parameter 
import matplotlib.pyplot as plt
x=[1,3,5,7,9]
y=[2,1,5,1,8]
plt.stem(x,y,linefmt=":",markerfmt="ro",basefmt="g",orientation="horizontal")
plt.show()
```

```python
#Basic boxplot
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x)
plt.show()
```

```python
#Parameters of boxplot
#To change the shape of boxplot to V shape we use notch parameter---->True or False
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,notch=True)
plt.show()
```

```python
#To chang to horizontal graph we use vert parameter
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,vert=False)
plt.show()
```

```python
#To change the width of the box----->Default width is 0.2----->widths parameter
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,vert=False,widths=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,widths=0.5)
plt.show()
```

```python
#To change the 1 in the x axis ----->Use tick_labels parameter and pass in list
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"])
plt.show()
```

```python
#patch_artist parameter------>To fill the color in the box ------>True or False
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True)
plt.show()
```

```python
#To show the mean in graph we use showmeans parameter 
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True)
plt.show()
```

```python
#Outlier
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70,300]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True)
plt.show()
```

```python
#to join with outlier use whis parameter
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70,300]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,whis=6.9)
plt.show()
```

```python
#We can also change the color and shape of the outler by sym parameter 
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70,200]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,sym="y+")
plt.show()
```

```python
#To hide it give blank
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70,200]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,sym=" ")
plt.show()
```

```python
#To change the properties of box we use boxprops
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,boxprops={"color":"r"})
plt.show()
```

```python
#To change the properties of cap we use capprops
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,boxprops={"color":"r"},capprops={"color":"r"})
plt.show()
```

```python
#To change the properties of wisk we use whiskerprops
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,boxprops={"color":"r"},whiskerprops={"color":"r"})
plt.show()
```

```python
#To chgange props of filer use flierprops
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70,200]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True,showmeans=True,boxprops={"color":"r"},capprops={"color":"r"},whiskerprops={"color":"r"},flierprops={"markeredgecolor":"g"})
plt.show()
```

```python
#Multiple boxplots
import matplotlib.pyplot as plt
x=[10,20,30,40,50,60,70]
y=[10,30,50,70,90]
l=[x,y]
plt.boxplot(l,tick_labels=["Python","Java"],patch_artist=True,showmeans=True)
plt.show()
```
