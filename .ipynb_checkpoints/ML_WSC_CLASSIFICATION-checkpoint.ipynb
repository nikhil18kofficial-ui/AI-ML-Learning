```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

dataset=pd.read_csv('decision_tree_pruning_dataset.csv')

```python
dataset.head(3)
```

```python
x=dataset.iloc[ : , :-1]
y=dataset['Purchased']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
for i in range(1,20):
    dc=DecisionTreeClassifier(max_depth=i)
    dc.fit(x_train,y_train)
    print(dc.score(x_train,y_train),dc.score(x_test,y_test),i)
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
dataset=pd.read_csv('knn_two_feature_dataset.csv')
```

```python
dataset.head(3)
```

```python
x=dataset.iloc[ : , :-1]
y=dataset['Purchased']
```

```python
sns.scatterplot(x='Age',y='EstimatedSalary',data=dataset,hue='Purchased')
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
ss.fit(x)
x=pd.DataFrame(ss.transform(x),columns=x.columns)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=3,p=1)
```

```python
knn.fit(x_train,y_train)
```

```python
knn.score(x_test,y_test)
```

```python
knn.score(x_train,y_train)
```

```python
#Check in loop for good k value 
for i in range(1,30):
    knn1=KNeighborsClassifier(n_neighbors=i)
    knn1.fit(x_train,y_train)
    print(knn1.score(x_train,y_train),knn1.score(x_test,y_test),i)
```

```python
#for decision boundary 
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=knn)
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
dataset=pd.read_csv('CGPA_vs_Package_CrossValidation.csv')
dataset.head(3)
```

```python
x=dataset.iloc[ : , :-1]
y=dataset['Package_LPA']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.model_selection import LeaveOneOut,LeavePOut,KFold
```

```python
lo=LeaveOneOut()
for train,test in lo.split(x_train,y_train):
    print(train,test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.model_selection import cross_val_score
x=cross_val_score(LinearRegression(),x_train,y_train,cv=5)
```

```python
x.sort()
x*100
```

```python
x=cross_val_score(LinearRegression(),x_train,y_train,cv=KFold())
```

```python
x.sort()
x*100
```
