```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from collections import Counter
from torch.utils.data import Dataset,DataLoader
from nltk import word_tokenize
import nltk
```

```python
def text_to_indeces(sentence,vocab):
    numerical_sentence=[]
    for token in sentence:
        if token in vocab:
            numerical_sentence.append(vocab[token])
        else:
            numerical_sentence.append(vocab['<unk>'])
    return numerical_sentence
```

```python
input_numerical_sentences=[]
for sentence in input_sentences:
    input_numerical_sentences.append(text_to_indeces(word_tokenize(sentence.lower()),vocab))
```

```python
len(input_numerical_sentences)
```

```python
#For sequences
training_sequence=[]
for sentence in input_numerical_sentences:
    for i in range(1,len(sentence)):
        training_sequence.append(sentence[:i+1])
    
```

```python
len_list=[]
for i in training_sequence:
    len_list.append(len(i))
```

```python
class CustomDataset(Dataset):
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __len__(self):
        return self.x.shape[0]
    def __getitem__(self,idx):
        return self.x[idx],self.y[idx]
```

```python
dataset=CustomDataset(x,y)
```

```python
dataloader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
for inpu,output in dataloader:
    print(inpu,output)
```

```python
class lstm_model(nn.Module):
    def __init__(self,vocab_size):
        super().__init__()
        self.embedding=nn.Embedding(vocab_size,100)
        self.lstm=nn.LSTM(100,150,batch_first=True)
        self.fc=nn.Linear(150,vocab_size)
    def forward(self,x):
        embeded=self.embedding(x)
        intermidiate_hidden,(final_hidden,final_cell)=self.lstm(embeded)
        final_hidden = final_hidden.squeeze(0)
        lin=self.fc(final_hidden)
        return lin
        
        
```

```python
loss_fn=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(params=model.parameters(),lr=0.001)
```

```python
epochs=50 
for epoch in range(epochs):
    t_loss=0
    X=0
    for x,y in dataloader:
        X+=1
        x,y=x.to(device),y.to(device)
        model.train()
        y_logits=model(x)
        
        train_loss=loss_fn(y_logits,y)
        t_loss+=train_loss
        optimizer.zero_grad()
        train_loss.backward()
        optimizer.step()
    t_loss/=X
    print(f"Epoch : {epoch} || loss: {t_loss}")
```

```python
def prediction(model,vocab,text):
    #Tokenize Text
    #text--->numerical indexes
    #Padding
    #Padded Text --->Model--->Output 
    #Predicted Word Index 
    #Fetch word by index 
    #Merge text 
    i=nltk.word_tokenize(text.lower())
    #print(i)
    x=text_to_indeces(i,vocab)
    #print(x)
    padding=torch.tensor([0]*(106-len(x))+x,dtype=torch.long).unsqueeze(0)
    #print(padding)
    #print(padding.shape)
    output=model(padding)
    #print(output)
    #print(output.shape)
    value,index=torch.max(output,dim=1)
    #print(index)
    #print(list(vocab.keys())[index])
    return text+' '+list(vocab.keys())[index]
```
