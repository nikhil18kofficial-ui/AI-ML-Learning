```python
def tokenize(text):
    text=text.lower()
    text=text.replace('?'," ")
    text=text.replace("'"," ")
    return text.split()
```

```python
vocab={'<UNK>':0}
```

```python
len(vocab)
```

```python
def build_vocab(row):
    tokenized_question=tokenize(row['question'])
    tokenized_output=tokenize(row['answer'])
    merged_token=tokenized_question+tokenized_output
    for token in merged_token:
        if token not in vocab:
            vocab[token]=len(vocab)
    
```

```python
dataset.apply(build_vocab,axis=1)
```

```python
vocab
```

```python
len(vocab)
```

```python
#Text to indices 
def text_to_indeces(text,vocab):
    indexed_text=[]
    for token in tokenize(text):
        if token in vocab:
            indexed_text.append(vocab[token])
        else:
            indexed_text.append(vocab['<UNK>'])
    return indexed_text
```

```python
text_to_indeces('What is nikhil',vocab)
```

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class QADataset(Dataset):
    def __init__(self,df,vocab):
        self.df=df
        self.vocab=vocab
    def __len__(self):
        return self.df.shape[0]

    def __getitem__(self,index):
        x=text_to_indeces(self.df.iloc[index]['question'],self.vocab)
        y=text_to_indeces(self.df.iloc[index]['answer'],self.vocab)
        return torch.tensor(x),torch.tensor(y)
```

```python
datasets=QADataset(dataset,vocab)
```

```python
datasets[10]
```

```python
datasets.__len__()
```

```python
dataloader=DataLoader(
    datasets,
    batch_size=1,
    shuffle=True
)
```

```python
for ques,answer in dataloader:
    print(ques,answer)
```

```python
import torch.nn as nn
```

```python
class SimpleRnn(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, 50)
        self.Rnn = nn.RNN(50, 64, batch_first=True)
        self.fc = nn.Linear(64, vocab_size)

    def forward(self, x):
        x = self.embedding(x)

        hidden, final = self.Rnn(x)

        final = final.squeeze(0)

        output = self.fc(final)

        return output
```

```python
model=SimpleRnn(len(vocab))
```

```python
loss_fn=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
ques, answer = next(iter(dataloader))

y_logits = model(ques)

print(ques.shape)
print(y_logits.shape)
print(answer.shape)
```

```python
epochs = 20

for epoch in range(epochs):
    total_loss = 0

    model.train()

    for ques, answer in dataloader:

        y_logits = model(ques)

        train_loss = loss_fn(
            y_logits,
            answer.squeeze(1)
        )

        optimizer.zero_grad()
        train_loss.backward()
        optimizer.step()

        total_loss += train_loss.item()

    print(f"Epoch: {epoch}, Loss: {total_loss}")
```
