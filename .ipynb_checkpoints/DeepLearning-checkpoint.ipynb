```python
#Single Layer Perceptron
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
dataset=pd.read_csv('perceptron_linearly_separable_100.csv')
```

```python
dataset.head(3)
```

```python
dataset['Placed'].value_counts()
```

```python
sns.scatterplot(x='CGPA',y='Aptitude_Score',data=dataset,hue='Placed')
plt.show()
```

```python
x=dataset.iloc[:,:-1]
y=dataset['Placed']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=100)
```

```python
from sklearn.linear_model import Perceptron
```

```python
pr=Perceptron()
```

```python
pr.fit(x_train,y_train)
```

```python
pr.score(x_train,y_train)*100,pr.score(x_test,y_test)*100
```

```python
#Single layer perceptron usually doesnot work perfectly
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=pr)
plt.show()
```
