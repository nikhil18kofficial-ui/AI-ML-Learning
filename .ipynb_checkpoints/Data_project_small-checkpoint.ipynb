```python
a=['Name', 'Gender', 'City', 'Department', 'Joining_Date', 'Rating']
for i in a:
    df[i].fillna(df[i].mode()[0],inplace=True)
```

```python
b=['Age', 'Salary', 'Experience']
from sklearn.impute import SimpleImputer
si=SimpleImputer()
for i in b:
    df[[i]]=si.fit_transform(df[[i]])
```

```python
c=['Age', 'Salary', 'Experience']
for i in c:
    z_score=(df[i]-(df[i].mean()))/(df[i].std())
    df=df[(z_score>-3) & (z_score<3)]
```
